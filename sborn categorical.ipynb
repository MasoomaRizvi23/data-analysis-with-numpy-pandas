```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
#---------------LOAD DATASET----------------------------------

sns.get_dataset_names()
penguins  = sns.load_dataset("penguins")

print(penguins)
```

```python
# Global Seaborn theme  

sns.set_context("notebook", font_scale=1.2)
sns.set_style("whitegrid")
sns.set_palette("muted")
plt.figure(figsize=(18,16))

# ---------------------------------------------------------
# 1. STRIPPLOT — Basic categorical scatterplot
# ---------------------------------------------------------  

plt.subplot(2, 2, 1)
sns.stripplot(
    data=penguins,
    x="species",
    y="bill_length_mm",
    jitter=True,          # spread points horizontally
    hue="sex",            # add second category
    dodge=True            # separate male/female
)
plt.title("Strip Plot: Bill Length by Species (with hue & dodge)")
plt.xlabel("Species")
plt.ylabel("Bill Length (mm)")



```

```python

sns.set_context("notebook", font_scale=1.2)
sns.set_style("whitegrid")
sns.set_palette("muted")
plt.figure(figsize=(18, 16))


# ---------------------------------------------------------
# 2. STRIPPLOT — No jitter (points overlap)
# --------------------------------------------------------- 

plt.subplot(2,2,2)
sns.stripplot(
    data = penguins,
    x = "species",
    y = "body_mass_g",
    jitter = False,
    hue = "sex",
    dodge = True
)
plt.title("Strip Plot: Body Mass by Species (jitter=False)")
plt.xlabel("Species")
plt.ylabel("Body Mass (g)")
```

```python
sns.set_context("notebook", font_scale=1.2)
sns.set_style("whitegrid")
sns.set_palette("muted")
plt.figure(figsize=(18, 16))



# ---------------------------------------------------------
# 3. SWARMPLOT — Smart non-overlapping points
# ---------------------------------------------------------
plt.subplot(2, 2, 3)
sns.swarmplot(
    data=penguins,
    x="species",
    y="flipper_length_mm",
    hue="sex",
    dodge=True
)
plt.title("Swarm Plot: Flipper Length by Species")
plt.xlabel("Species")
plt.ylabel("Flipper Length (mm)")
```

```python






# ---------------------------------------------------------
# 4. SWARMPLOT — Custom order on x-axis
# ---------------------------------------------------------
plt.subplot(2, 2, 4)
sns.swarmplot(
    data=penguins,
    x="species",
    y="bill_depth_mm",
    order=["Gentoo", "Adelie", "Chinstrap"],   # custom order
    hue="sex",
    dodge=True
)
plt.title("Swarm Plot: Bill Depth (Custom Order)")
plt.xlabel("Species")
plt.ylabel("Bill Depth (mm)")

plt.tight_layout()
plt.show()
```
