```python
#splitting the line in dataabs
data = []

with open("csv0.csv") as fin:
  for line in fin:
    data.append(line.strip().split(','))
print(data)
```

```python
#creating new filen
with open("csv0.csv", "w") as f:
    f.write("NAME,AGE,CITY\n")
    f.write("masooma, 36, oakville\n")
    f.write("ali, 8, oakville\n")
    f.write("abrahm, 6, oakville\n")
  
```

```python
#adding more row at the bottom "appending"
with open("csv0.csv","a") as f:
    f.write("mom, 60, lucknow\n")
    f.write("dad, 65, lucknow\n")
```

```python
#reader( read the line)
import csv

data = []
with open("csv0.csv") as fin:
    reader = csv.reader(fin)
    for row in reader:
        data.append(row)
print(data)
```

```python
#dictreader (read data in dictionary)
import csv

with open("csv0.csv") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
```

```python
#restval..use when row has missing value

with open("csv0.csv") as f:
    reader = csv.DictReader(f, restval="UNKNOWN")
    for row in reader:
        print(row)
```
