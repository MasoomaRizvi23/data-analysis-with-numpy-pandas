```python
import re

# Mapping of written numbers to digits
word_to_num = {
    "zero":0, "one":1, "two":2, "three":3, "four":4, "five":5,
    "six":6, "seven":7, "eight":8, "nine":9, "ten":10,
    "eleven":11, "twelve":12, "thirteen":13, "fourteen":14,
    "fifteen":15, "sixteen":16, "seventeen":17, "eighteen":18,
    "nineteen":19, "twenty":20, "thirty":30, "forty":40,
    "fifty":50, "sixty":60, "seventy":70, "eighty":80,
    "ninety":90
}

def convert_word_age(text):
    text = text.lower().strip()
    if text in word_to_num:
        return word_to_num[text]
    return None

ages = []

with open(r"C:\Users\masri\Downloads\6_6_7_Wrap_Up_messy2.csv", "r", encoding="utf-8", errors="ignore") as f:
    for line in f:
        line = line.strip()

        # 1. Extract numeric ages
        nums = re.findall(r"\d+", line)
        nums = [int(n) for n in nums if 0 <= int(n) <= 120]

        # 2. Extract word ages (like "thirty")
        words = re.findall(r"[A-Za-z]+", line)
        for w in words:
            w_age = convert_word_age(w)
            if w_age is not None and 0 <= w_age <= 120:
                nums.append(w_age)

        # 3. If any valid age found, take the FIRST one
        if nums:
            ages.append(nums[0])

average_age = sum(ages) / len(ages)
average_age
```
