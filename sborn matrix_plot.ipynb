```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

sns.get_dataset_names()
df_mpg  = sns.load_dataset("mpg")
print(df_mpg)
#================================================================================== 

#CORELATIONAL MAP
#Get only numerc data from the DataFrame
#================================================================================== 

sns.heatmap(df_mpg.pivot_table("mpg","model_year","origin","mean"),annot=True,cmap="coolwarm", fmt=".2f")
```

```python
#================================================================================== 

#CORELATIONAL MAP
#Get only numerc data from the DataFrame
#================================================================================== 

df_numeric_mpg = df_mpg.drop(["origin", "name"], axis=1)
correlation_matrix = df_numeric_mpg.corr()

plt.figure(figsize=(12,10))
sns.heatmap(correlation_matrix,  annot=True, 
               cmap="coolwarm", fmt=".2f")
```

```python
#================================================================================== 

#CLUSTER MAP
#================================================================================== 

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd




#Get only numeric data from the DataFrame
df_numeric_mpg = df_mpg.drop(["origin", "name"], axis=1)
correlation_matrix = df_numeric_mpg.corr()

sns.clustermap(correlation_matrix,  annot=True, 
               cmap="coolwarm", fmt=".2f", figsize=(10, 8))
```
