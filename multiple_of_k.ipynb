```python
import numpy as np
def multiple_of_k(n:int, k:int) -> np.ndarray[int]:
    """
returns all multiples of k that are less than or equal to n.
requires: n>0
    Examples:
     multiples_of_k(1, 1) => np.ndarray([0,1])
     multiples_of_k(5, 2) => np.ndarray([0,2,4])
     multiples_of_k(6, 3) => np.ndarray([0,3,6])
    """
    return np.arange(0,n+1,k)

print(multiple_of_k(10,2))
```
