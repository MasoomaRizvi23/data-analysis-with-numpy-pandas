```python
import pytest
import ipytest
ipytest.autoconfig() 
ipytest.run()

```

```python

def compare_strings(str1, str2, case_sensitive = True):
        if not case_sensitive:
            str1 = str1.lower()
            str2 = str2.lower()
        result = str1 == str2
        return result

```

```python
%%ipytest


def test_compare_strings():
    test_string1 = "Ontario"
    test_string2 = "ontario"
    expected_reult = False
    real_result = compare_strings(test_string1, test_string2)
    assert expected_reult == real_result


```

```python
%%ipytest
ipytest.clean()
@pytest.mark.parametrize("test_string1, test_string2, expected_result", [
    ("Ontario", "ontario", False),  # Test False due to case sensitive
    ("Ontario", "Ontario", True),   # Test True when case sensitive
])
def test_compare_strings_param(test_string1, test_string2, expected_result):
    real_result = compare_strings(test_string1, test_string2)
    assert (real_result == expected_result)
```

```python
%%ipytest
@pytest.mark.parametrize("test_string1, test_string2, case_arg, expected_result", [
    ("Ontario", "ontario", True, False),   #Test False due to case sensitive
    ("Ontario", "Ontario", True, True),    #Test True when case sensitive
    ("Ontario", "ontario", False, True),   #Test True when case insensitive strings compared
    ("Ontario", "Ontario", False, True),   #Test True when equal and case insensitive
])

def test_compare_strings_params_case(test_string1, test_string2, case_arg, expected_result):
    real_result = compare_strings(test_string1, test_string2, case_arg)
    assert (real_result == expected_result)
```
