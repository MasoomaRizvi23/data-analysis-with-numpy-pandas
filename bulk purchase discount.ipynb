```python
import numpy as np
def total_cost(products:np.ndarray[int],
              threshold:np.ndarray[int],
              discounted_price:np.ndarray[int],
              default_price: np.ndarray[int]) -> np.ndarray[int]:
    """
  Returns the cost for each company in products. Cost is computed
  using the discounted_price if the item exceeds the threshold and
  the default_price for the item otherwise.
  
  Requires:
     len(products) > 0
     len(products.shape) == 2
     products.shape[1] == len(thresholds) 
     products.shape[1] == len(discounted_price) 
     products.shape[1] == len(default_price)
     
  Example:
     total_cost(np.array([[5, 10], 
                          [3, 7]]), 
                np.array([6, 6]), 
                np.array([4, 5]), 
                np.array([8, 9])) => np.array([90, 59])
    """
#boolean mask: True where quantity>= threshold, False otherwise.abs
    use_discount = products>=threshold

#compute cost per item using vectorized operations.

    cost_matrix = np.where(use_discount, products*discounted_price, products*default_price)
    return cost_matrix.sum(axis =1)
    
#test

products = np.array([[5, 10],
                     [3, 7]])

thresholds = np.array([6, 6])
discounted_price = np.array([4, 5])
default_price = np.array([8, 9])
result = total_cost(products, thresholds, discounted_price, default_price)
print(result)
```
