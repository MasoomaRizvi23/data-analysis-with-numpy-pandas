```python
import re

print("Example 1 AM and PM:")
pattern = "((0[1-9])|(1[0-2])):[0-5][0-9] *(A|P)M"
target = "The time is 01:45AM, not 10:45 PM!"
print(re.search(pattern, target))

print("Example 2 AM and PM:")
pattern = "((0[1-9])|(1[0-2])):[0-5][0-9] +(A|P)M"
target = "The time is 01:45AM, not 10:45 PM!"
print(re.search(pattern, target))
```

```python
import re

with open ("pride_and_prejudice.txt", "w") as f:
    f.write("PRIDE and PREJUDICe by Jane Austen")
    
with open("pride_and_prejudice.txt", 'r') as fin:
  book = fin.read()
print(book) #Print title of Pride and Prejudice
```

```python
book = """
Darcy is a person.
This is a line not containing the name of the character.
This is Darcy's apple.
The word Darcy occurs here and Darcy occurs here as well.
I have a feeling *Darcy* was here.
"""
```

```python
# Find all instances of the characters Darcy
matches_Darcy = re.findall("Darcy", book)
print("number of occurence of word daecy", len(matches_Darcy))
print("Matches found:", matches_Darcy)
```

```python

matches_Darcy = re.findall(".*Darcy*.", book)
print("number of occurence of word daecy", len(matches_Darcy))
print("Matches found:", matches_Darcy)
```

```python
# Find all instances of the or The
matches_the = re.findall("[tT]he", book)

print(f"Number of occurrences of lines containing The or the: {len(matches_the)}")
print("Matches found:", matches_the)
```

```python
# An alternate way(pun intended!) Note concatenation takes
# precedence over alternation - more on this later.
matches_the_alt = re.findall("the|The", book)

print(f"Number of occurrences of lines containing The or the: {len(matches_the_alt)}")
print("Matches found:", matches_the_alt)
```

```python
# Containing Th. but not counting the
matches_th_not_the = re.findall(".*[tT]h[^e]", book)   
print(f"Number of occurrences of lines containing th. or Th. "
      f"without counting the or The: {len(matches_th_not_the)}")
print("Matches found:", matches_th_not_the)
```
