```python
import matplotlib.pyplot as plt 
import pandas as pd 
import seaborn as sns 


penguins  = sns.load_dataset("penguins")
print(penguins)


plt.figure(figsize=(6,15))
plt.subplot(3,1,1)
sns.boxplot(penguins, x="species", y="bill_depth_mm", 
            hue="sex", palette="Set3")
plt.title("Box Plot: Bill Depth Across Species and Sex")

plt.subplot(3,1,2)
sns.boxenplot(penguins, x="species", y="body_mass_g", hue="island")
plt.title("Boxen Plot: Body Mass Across Species")

plt.subplot(3,1,3)
sns.violinplot(penguins, x="species", y="flipper_length_mm", hue="sex")
plt.title("Violin Plot: Flipper Length Across Species")

plt.show()
```
