```python
import matplotlib.pyplot as plt
import numpy as np 

x = np.linspace(0,10,100)
y= np.sin(x)

plt.figure(figsize = (10,10))   #need to rotate the size
plt.plot(x,y,label = "Sine Wave")

plt.title("Sine Wave", fontsize = 10, family = "serif", fontweight="bold", rotation = -20,color = "darkblue")

plt.xlabel("X-axis", fontsize=14, fontweight='normal', 
              family='sans-serif', color='green')
plt.ylabel("Y-axis", fontsize=14, fontweight='light', 
              family='monospace', color='red', alpha = 0.5)
plt.legend(loc = "lower right",prop={"size":12, 
                "family":'serif', 
                "weight":'bold'})
plt.show()
```

```python
#SOME MORE PROPERTIES OF CUSTOMIZATION

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(10, 10))
plt.plot(x, y, label="Sine Wave")

plt.title("Sine Wave Plot", 
            fontsize=18, 
            fontweight='bold', 
            fontstyle='italic', 
            family='serif', 
            color='darkblue',
            rotation=-20)
            
plt.xlabel("X-axis", fontsize=14, fontweight='normal', 
              family='sans-serif', color='green')
plt.ylabel("Y-axis", fontsize=14, fontweight='light', 
              family='monospace', color='red', alpha = 0.5)
              
plt.legend(loc='lower right', 
          prop={"size":12, 
                "family":'serif', 
                "weight":'bold'})


plt.text(5, 0.5, 'Text at (5, 0.5)', 
        fontsize=14,fontweight='light', fontstyle='italic', family='serif', 
        color='brown', ha='center', va='top', alpha=0.7)

       
plt.annotate(r'Peak of $\sin(x)$', 
            xy=(np.pi/2, 1), xytext=(np.pi+1, 0.5),
            fontsize=14, fontweight='bold', fontstyle='italic', 
            family='serif', color='darkblue', rotation=30,
            ha='center', va='bottom',
            arrowprops=dict(facecolor='black', arrowstyle='fancy'),
            bbox=dict(facecolor='yellow', edgecolor='red', boxstyle='round,pad=0.5'),
            label='Peak Annotation',
            annotation_clip=True,
            url="https://watspeed.uwaterloo.ca/index.html"
)

#If using at home, save it as an svg to see the url link
plt.savefig("plot.svg")
plt.show()
```
