```python
import numpy as np
import pandas as pd

data = {
    'Test1': [87, 92, 76, 95, 88],
    'Test2': [90, 85, 78, 94, 84],
    'Test3': [91, 89, 80, 97, 85]
}

index = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']

df = pd.DataFrame(data, index=index)

print(df)

#Percentages (apply also works here)
print(df.map(lambda x: x/100))


def get_grade(score):
    if score >= 80:
        return 'A'
    elif score >= 70:
        return 'B'
    elif score >= 60:
        return 'C'
    elif score >= 50:
        return 'D'
    else:
        return 'F'

#Getting letter grades 
# Note apply will not work here as it tries to apply this column wise
# which will end up in the NumPy errors from earlier in the course
print(df.map(get_grade))

#A more advanced way to get the grades correct. Don't worry if you don't
#understand exactly how this works with the generators and next command.
#I thought this was cool when I discovered this and wanted to share!
grade_map = {range(80, 101): 'A', range(70, 81): 'B', range(60, 71): 'C', 
             range(50, 61): 'D', range(0, 50): 'F'}
print(df.map(lambda x: next((grade for score_range, grade in grade_map.items() 
                             if x in score_range), None)))

#Heat map of "high scores"
category_map = {True: 'High', False: 'Low'}
print(df.map(lambda x: category_map[x > 85]))

#Mapping Groups
name_map = pd.Series({'Alice': 'Group A', 
                      'Bob': 'Group B',
                      'Charlie': 'Group A', 
                      'David': 'Group A', 
                      'Eve': 'Group B'})
df['Group'] = df.index.map(name_map)
print(df[['Test1', 'Test2', 'Test3', 'Group']])
```
