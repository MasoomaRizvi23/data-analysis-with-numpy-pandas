```python
import numpy as np 
import pandas as pd    

df = pd.read_csv(r"C:\Users\masri\Downloads\CSV_FILES_for data munging\module6_1_data.csv",encoding = "latin1", comment = "#")
print(df["Age"].value_counts(dropna = False))
print(df["Income"].value_counts(dropna = False))
print(df["Score"].value_counts(dropna = False))
print("\n\n-----------------------------------\n\n")

#replace nan with -1
df["Age"] = df["Age"].replace(np.nan, -1).astype(int)
df["Income"] = df["Income"].replace("-", "-1").astype(int)


#how many duplicate data are there
print(df.duplicated().sum())                                                     #IDENTIFY DUPLICATES
print("\n\n--------------------------------------\n\n")


#how to find dupicate data                                                      #VIEW DUPLICATES
print(df.duplicated())
print("\n\n------------------------------------------------\n\n")

      
#how to count duplicate data
print(df.duplicated().value_counts())
print("\n\n-----------------------------------\n\n")

#drop duplicate "data 
print(df.drop_duplicates())                                                 #REMOVE DUPLICATES
print("\n\n--------------------------------------\n\n")
#reset the index after removing duplicates 

print(df.reset_index(drop = True))         #RESET INDEX
print("\n\n---------------------------------------------\n\n")
```
