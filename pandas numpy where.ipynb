```python
#IMORTING LIBRARIES
import numpy as np 
import pandas as pd    
```

```python
#Creating Two Series
s1 = pd.Series([1,5,3])
s2 = pd.Series([3,1,9])
```

```python
# USING WHERE 
print("where()")
```

```python
#using where() as numpy
# we will convert the series into array 

A = np.array(s1)
 #using numpy where version 

print("using numpy where version")
print(np.where(A>2, 99,-1))
```

```python
# using where() as pandas \
print("using pandas where version") 
print(s1.where(s1>2))
```

```python
#pandas where() with a Replacement Value
python
print(s1.where(s1>2, other = -2))
```

```python
#combine section 
print(s1.combine(s2,min))
print(s1.combine(s2,max))
```

```python
# NumPy Minimum and maximum

print(np.minimum(s1,s2))
print(np.maximum(s1, s2))
```
