```python
import numpy as np

print("Comparing different data types for the same array")
arr_int = np.array([1,2,3])
arr_complex = np.array([1,2,3], dtype=complex)
arr_float = arr_int.astype(float)
arr_str = arr_int.astype(str)
print("Integer Array")
print(arr_int)
print("Complex Number Array")
#Don't worry if you haven't seen a complex number before!
print(arr_complex)
print("Float Array")
print(arr_float)
print("Str Array")
print(arr_str)
print("The int64 dtype")
print(arr_int.dtype)
```
