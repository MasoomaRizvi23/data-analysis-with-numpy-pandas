```python
import matplotlib.pyplot as plt 
import numpy as np 

x = np.linspace(0,10,10)
plt.figure(figsize=(10,10))
plt.plot(x, np.sin(x), linestyle = "-", linewidth = 2, color = "b", marker = "o",markersize = 2, label = "solid circle line")
plt.plot(x, np.cos(x), "r--^", label = "dashed triangle line")
plt.plot(x,np.tan(x)/10, "g:s", markersize = 2, label = "dotted square line")
plt.plot(x, np.exp(-x/3), "m:D", markersize=6, label='Dash-dot Diamond Line')

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title('Example of Different Linestyles, Linewidths, Colors, and Markers')
plt.legend()
plt.show()
```
