```python
import numpy as np

def total_cost(ages: np.ndarray[int]) -> int:
    """
  Returns the total cost of all patrons with
  ages given in the array according to the pricing scheme
  
  Example:
    total_cost(np.array([1, 10, 20, 80])) => 23
    """
        def ticket_price(age):
            if age < 5:
                return 0
            elif  age < 18:
                return 5
            elif  age >= 65:
                return 8
            else:
                return 10
        prices = np.vectorize(ticket_price) 
        return int(np.sum(prices(ages)))

# Test
print(total_cost(np.array([1, 10, 20, 80])))  # 23
```
