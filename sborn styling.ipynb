```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd


iris = sns.load_dataset("iris")
iris.head()
plt.figure(figsize=(20,20))
plt.subplot(2,2,1)
sns.scatterplot(iris)

# 1. Control sizes (fonts, lines, markers)
sns.set_context("notebook", font_scale=1.2)

# 2. Control background, grids, ticks
sns.set_style("whitegrid")

# 3. Control colors
sns.set_palette("muted")
```

```python
iris = sns.load_dataset("iris")

plt.figure(figsize=(10,6))
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species")
plt.title("Sepal Length vs Width by Species")
plt.show()
```

```python
with sns.axes_style("ticks"):
    sns.kdeplot(data=iris, x="petal_length", hue="species", fill=True)
    plt.title("Petal Length KDE")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Global theme
sns.set_context("notebook", font_scale=1.2)
sns.set_style("whitegrid")
sns.set_palette("muted")

iris = sns.load_dataset("iris")

plt.figure(figsize=(8,10))
plt.axis('off')
plt.title('KDEs from Iris Dataset')

# 1. KDE of one column
plt.subplot(3,1,1)
sns.kdeplot(iris['sepal_length'], fill=True)

# 2. KDE of all numeric columns
plt.subplot(3,1,2)
sns.kdeplot(iris[iris.columns[:-1]], fill=True)

# 3. KDE separated by species
plt.subplot(3,1,3)
sns.kdeplot(data=iris, x="sepal_length", hue="species", fill=True)

plt.subplots_adjust(hspace=0.5)
plt.show()
```
