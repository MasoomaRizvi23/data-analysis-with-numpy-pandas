```python
import matplotlib.pyplot as plt
import numpy as np 

def draw_x():

    plt.plot([0.2,0.8],[0.2,0.8], color = "blue", linewidth = 3)
    plt.plot([0.2,0.8],[0.8,0.2], color = "blue", linewidth = 3)

def draw_o(): 
     # Using polar coordinates to manually draw circle   
    middle = (0.5,0.5)
    radius = 0.4
    theta = np.linspace(0,2*np.pi, 200)

    x = middle[0]+radius*np.cos(theta)
    y = middle[1]+radius*np.sin(theta)

    plt.plot(x,y,color = "green", linewidth=3)

def draw_move(index, symbol):

    plt.subplot(3,3,index)
    ax = plt.gca()  # get current axes
    if symbol == "X": 
        ax.text(0.5,0.75, f"{index}", ha = "center", va = "center", fontsize = 24, color = "purple")
        draw_x()
    elif symbol == "O": 
        ax.text(0.5, 0.5, f'{index}', ha= "center", va ="center", fontsize = 24, color = "purple")
        draw_o()
    #remove axes 
    plt.xticks([])
    plt.yticks([])
    plt.xlim(0,1)
    plt.ylim(0,1)
    plt.axis("off")


#------------------------------------------------RUNNING THE GAME-----------------------------

plt.figure(figsize=(6, 6))

#Move 1: Draw X in middle
draw_move(5, 'X')
plt.savefig("move1.png")

#Move 2: Draw O in top right
draw_move(3, 'O')
plt.savefig("move2.png")

#Move 3: Draw X in top middle
draw_move(2, 'X')
plt.savefig("move3.png")

#Move 4: Draw O in bottom middle
draw_move(8, 'O')
plt.savefig("move4.png")

#Move 5: Draw X in bottom right
draw_move(9, 'X')
plt.savefig("move5.png")

#Move 6: Draw O in top left
draw_move(1, 'O')
plt.savefig("move6.png")

#Move 7: Draw X in middle right
draw_move(6, 'X')
plt.savefig("move7.png")

#Move 8: Draw O in middle left
draw_move(4, 'O')
plt.savefig("move8.png")

#Move 9: Draw X in bottom left
draw_move(7, 'X')
plt.savefig("move9.png")
```
