```python
import csv
def triple_doubles(fnamein:str) -> list[str]:

    """ 
    consume a filename an return list of players who have earned the triple doible.
    REQUIRES: fnamein a valid csv file
    EXAMPLE:triple_double("empty.csv") => []
    assuming empty.csv is empty
    
    assuming ex.csv contains
    Name,Points,Rebounds,Assists,Steals,Blocks
    LeBron James,30,11,10,2,1
    Shaquille O'Neal,28,15,2,0,5
    Allen Iverson,35,3,7,3,0
    Kevin Durant,32,9,5,1,2
    Tim Duncan,22,12,4,1,4
    Dirk Nowitzki,27,10,3,1,1
    Chris Paul,15,4,13,4,0
    James Harden,44,7,10,3,1
    Russell Westbrook,18,10,10,2,1
    Stephen Curry,41,6,9,5,0
    
    """
    result = []
    with open(fnamein, newline='') as f:
        reader = csv.DictReader(f)
        for row in reader:
            stats = [
                int(row["Points"]),
                int(row["Rebounds"]),
                int(row["Assists"]),
                int(row["Steals"]),
                int(row["Blocks"])
            ]

            if sum(stat >= 10 for stat in stats) >= 3:
                result.append(row["Name"])

    return sorted(result)
```

```python
csv_data = """Name,Points,Rebounds,Assists,Steals,Blocks
LeBron James,30,11,10,2,1
Shaquille O'Neal,28,15,2,0,5
Allen Iverson,35,3,7,3,0
Kevin Durant,32,9,5,1,2
Tim Duncan,22,12,4,1,4
Dirk Nowitzki,27,10,3,1,1
Chris Paul,15,4,13,4,0
James Harden,44,7,10,3,1
Russell Westbrook,18,10,10,2,1
Stephen Curry,41,6,9,5,0
"""
with open("ex_test.csv", "w") as f:
    f.write(csv_data)
```

```python
%%ipytest
import pytest

def test_triple_doubles_output():
    expected=['LeBron James', 'Russell Westbrook']
    assert triple_doubles("ex_test.csv") == expected

def test_no_triple_doubles_in_empty_file(tmp_path):
    empty = tmp_path / "empty.csv"
    empty.write_text("Name,Points,Rebounds,Assists,Steals,Blocks\n")
    assert triple_doubles(str(empty)) == []

def test_single_triple_doubles(tmp_path):
    data = """Name,Points,Rebounds,Assists,Steals,Blocks
John Doe,10,10,10,1,1
"""

    file = tmp_path/"one_csv"
    file.write_text(data) 
    assert triple_doubles(str(file)) ==["John Doe"]
```
