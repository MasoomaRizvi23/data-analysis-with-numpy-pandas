```python
#showing pie chart for all quizez

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

def convert_to_grade(score):
    if score < 50:
        return 'F'
    elif score < 60:
        return 'D'
    elif score < 70:
        return 'C'
    elif score < 80:
        return 'B'
    else:
        return 'A'

letters = dfq.drop("Name", axis=1).map(convert_to_grade)
plt.figure(figsize=(16,8))
for i in range(5):
    plt.subplot(2,4,i+1)
    grade_counts = letters[f"Quiz {i+1}"].value_counts()

    colors = plt.cm.Set2(range(len(grade_counts)))
    
    plt.pie(grade_counts, labels=grade_counts.index, 
        autopct='%.2f%%', startangle=90, colors=colors,
       textprops={"size":14})
    plt.title(f"Quiz {i+1}")

plt.tight_layout(rect=[0, 0, 1, 0.95]) 
plt.suptitle('Distribution of Letter Grades All Quizzes')

plt.show()
```
