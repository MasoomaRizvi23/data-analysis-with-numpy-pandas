```python

            
import pytest
import ipytest
import re
ipytest.autoconfig()
ipytest.clean()



def foo1(password):
    if len(password) < 8 or len(password) > 20:
        return False
    if any(c.isupper() for c in password):
        return True
    if any(c.islower() for c in password):
        return True
    if any(c.isdigit() for c in password):
        return True
    if any(c in "!@#$%^&*()-+" for c in password):
        return True
    return False

def foo2(password):
    if len(password) <= 8 or len(password) >= 20:
        return False
    has_upper = any(c.isupper() for c in password)
    has_lower = any(c.islower() for c in password)
    has_digit = any(c.isdigit() for c in password)
    has_special = any(c in "!@#$%^&*()-+" for c in password)
    return has_upper and has_lower and has_digit and has_special

def foo3(password):
    if not (8 <= len(password) <= 20):
        return False
    if not re.search(r"[A-Z]", password):
        return False
    if not re.search(r"[a-z]", password):
        return False
    if not re.search(r"\d", password):
        return False
    if not re.search(r"[!@#\$]", password):
        return False
    return True

def foo4(password):
    if not (8 <= len(password) <= 20):
        return False
    if not re.search(r"[A-Z]", password):
        return False
    if not re.search(r"[a-z]", password):
        return False
    if not re.search(r"\d", password):
        return False
    if not re.search(r"[!@#$%^&*()\-\+]", password):
        return False
    return True
    

def foo5(password):
    if not (8 <= len(password) <= 20):
        return False
    if not re.search(r"[A-Z]", password):
        return False
    if not re.search(r"[a-z]", password):
        return False
    if not re.search(r"[!@#$%^&*()\-\+]", password):
        return False
    return True

def foo6(password):
    if len(password) < 8 or len(password) > 20:
        return False
    has_upper = any(c.isupper() for c in password)
    has_lower = any(c.islower() for c in password)
    has_digit = any(c.isdigit() for c in password)
    has_special = any(c in "!@#$%^&*()-+" for c in password)
    return has_upper or has_lower or has_digit or has_special


tests_per_foo =  [

            # ✅ VALID passwords
    ["Hello1!@", True],               # exactly 8 chars (min boundary)
    ["Hello1!@12345678901", True],    # exactly 20 chars (max boundary)
    ["Hello1!@abc", True],            # normal valid password
    ["ValidPass1!", True],            # uses ! symbol
    ["ValidPass1#", True],            # uses # symbol
    ["ValidPass1%", True],            # uses % symbol
    ["ValidPass1^", True],            # uses ^ symbol
    ["ValidPass1&", True],            # uses & symbol
    ["ValidPass1*", True],            # uses * symbol
    ["ValidPass1(", True],            # uses ( symbol
    ["ValidPass1)", True],            # uses ) symbol
    ["ValidPass1-", True],            # uses - symbol
    ["ValidPass1+", True],            # uses + symbol

    # ❌ LENGTH FAILURES
    ["Hi1!", False],                  # too short (4 chars)
     ["Hello1!",False],              # 7 chars, one below minimum
    ["Hello1!@1234567890123", False],  # 21 chars, one above maximum
    ["", False],                      # empty string

    # ❌ MISSING UPPERCASE
    ["hello1!@", False],              # no uppercase

    # ❌ MISSING LOWERCASE
    ["HELLO1!@", False],              # no lowercase

    # ❌ MISSING DIGIT
    ["HelloABC!@", False],            # no digit

    # ❌ MISSING SYMBOL
    ["Hello123ab", False],            # no special character

    # ❌ BOUNDARY: exactly 8 but missing a requirement
    ["hello12!", False],              # 8 chars but no uppercase
    ["HELLO12!", False],              # 8 chars but no lowercase
    ["HelloABC!", False],             # 8 chars but no digit
    ["Hello123", False],              # 8 chars but no symbol

    # ❌ BOUNDARY: exactly 20 but missing a requirement
    ["hello1!@bcdefghijkl", False],   # 20 chars, no uppercase
    ["HELLO1!@BCDEFGHIJKL", False],   # 20 chars, no lowercase
    ["HelloABC!@defghijkl", False],   # 20 chars, no digit
    ["Hello123abcdefghijk", False],   # 20 chars, no symbol
  
]
            
          

functions = [foo1, foo2, foo3, foo4, foo5, foo6]
tests = [[f] + L for f in functions for L in tests_per_foo]

@pytest.mark.parametrize("f, s, expected_result", tests)
def test_all_foo(f, s, expected_result):
    assert (f(s) == expected_result), \
      f"{f.__name__}('{s}') returned {f(s)}, expected {expected_result}"

          
ipytest.run()
```
