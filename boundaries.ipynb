```python
import re
text = "cat\ncatalogue\nconcatenate\nscat"

#simple find all cat matches 
pattern = "cat"
print("match1", re.findall(pattern, text)) 
```

```python
# 2. Match exact string "cat" (no multiline, no match at
#  beginning or end of lines)
pattern_exact = r"^cat$"
print("match2", re.findall(pattern_exact, text))
```

```python
# 3. Match exact string "cat" at the beginning and end of lines using MULTILINE 
print("match3", re.findall(pattern_exact, text, re.MULTILINE)) 
```

```python
#Match "cat" as a word boundary match
pattern_exact1 = r"\bcat\b"
print("match4", re.findall(pattern_exact1, text)) 
```

```python
 #Match "cat" as a word boundary at the beginning of a word
pattern_exact2 = r"\bcat"
print("match5", re.findall(pattern_exact2, text))  
```

```python
#Match "cat" at the end of a string
pattern_exact3 = r"cat$"
print("match6", re.findall(pattern_exact3, text)) 
```

```python
#Match "cat" as a word boundary at the end of a word
pattern_exact4 = r"cat\b"
print("match7", re.findall(pattern_exact4, text))  
```

```python
#Match "cat" as interior (not at word boundary)
pattern_interior = r"\Bcat\B"
print("match8", re.findall(pattern_interior, text)) 
```

```python
# Match 5 numbers
pattern_numbers = r"\b\d{5}\b"
s = "Matches things like 12345\nBut not 123456"\
    "\nor 1234\n however 44330 works!"
print(re.findall(pattern_numbers, s, re.MULTILINE))
```

```python
#matching an ID 
pattern_ID = r"^[a-zA-Z]{2}[0-9]{2,4}$"
s = """
AB123\ncd456\nxy789\nxy78\nxy789324
hello 123\nABCD567\nzz321\nat1234
test abc 999
"""
print("pattern_ID", re.findall(pattern_ID, s, re.MULTILINE))
```

```python
import re
pattern = r"^(\d{3})-(\d{3})-(\d{4})$"
s = """
123-456-7890
1234-567-8901
12-345-6789
519-888-4567
123-45-6789
abc-123-4567
123-abc-7890
987-654-3210
123-4567-8901
123-456-789
123-456-78900
"""
for m in re.finditer(pattern , s , re.MULTILINE):
    print("........")
    print("phone number:", m.group())
    print("phone number Area code", m.group(1))
    print("phone number Telephone prefix:", m.group(2))
    print("phone number Line Number:", m.group(3))
```

```python
import numpy
import re
pattern2 = re.compile(r"""
    (\d{3})   # First three number group
    -         # match dash
    (\d{2})   # Second two number group
    -         # match dash
    (\d{4})   # Third four number group            
""", re.MULTILINE|re.VERBOSE)

text = "123-45-6789\n235-435-6789"
for m2 in pattern2.finditer(text):
    if m2: 
            #Note m2.group(0) also returns full match.
        print("Full match:", m2.group())         # Output: Full match: 123-45-6789
        print("First part:", m2.group(1))        # Output: First part: 123
        print("Second part:", m2.group(2))       # Output: Second part: 45
        print("Third part:", m2.group(3))        # Output: Third part: 6789
```

```python
##REGULAR EXPRESSION WITH NUMPY BY USING VECTORIZATION

import numpy as np
import re

arr = np.array(['apple', 'banana', 'cherry', 'apricot', 'pineapple'])

def starts_with_a(s):
    return bool(re.search(r'^a', s))  

vectorized_check = np.vectorize(starts_with_a)
mask = vectorized_check(arr)

print(arr[mask])


def replace_a_with_X(s):
    return re.sub(r'a', 'X', s)

vectorized_replace = np.vectorize(replace_a_with_X)
result = vectorized_replace(arr)

print(result)
```
