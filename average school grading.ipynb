```python
import pandas as pd

def average(df: pd.DataFrame) -> float:
    '''
  Returns the average of the df["Grade"] column
  where letters are converted to numeric types,
  FTC, NMR, DNW, WF are all 32s,
  numeric types are left alone and all other values 
  are dropped from the computation
  
  Requires: 
    df.shape == (n, 2)
    list(df.columns) == ["Name", "Grade"]
    df["Grade"] is a string integer between 0 and 100
      or one of the valid acronyms above.
    
  Examples:
    average(pd.DataFrame({})) => 0.0
    average(pd.DataFrame({"Name":["Alice", "Bob"], 
                          "Grade": ["90", "70"]})) => 80.0
    '''
    #if DataFrame is empty return 0.0

    if df.empty:
        return 0.0

    #mappig for letter grades

    letter_map = {

        "A":89,
        "B":75,
        "C": 65,
        "D": 55,
        "F": 38
    }

    #codes that should be dropped 
    drop_codes = {"INC", "MM", "WD","CR","NCR","",None}

    #code that should be treated as 32 
    penalty_codes = {"FTC","NMR","DNW","WF"}

    numeric_values = []
    for grade in df["Grade"]:
        grade = str(grade).strip()     #ensures its a clean string

        if grade in drop_codes:
            continue
        if grade in penalty_codes: 
            continue
        if grade in letter_map:
            numeric_values.append(letter_map[grade])
            continue
        if grade.isdigit():
            numeric_values.append(int(grade))
            continue
        continue      # If something unexpected appears, ignore it # (safe fallback)

      # If no valid numeric values remain, return 0.0
    if len(numeric_values) == 0:
        return 0.0

    # Compute and return the average
    return sum(numeric_values) / len(numeric_values)

df = pd.DataFrame({
    "Name": ["Alice", "Bob"],
    "Grade": ["90", "70"]
})

print(average(df))
```
