```python
import pandas as pd 
df = pd.read_csv(r"C:\Users\masri\Downloads\4_8_df_classical.csv")
df
```

```python
# How many rows are in data frame
print(df.shape)
df.duplicated().sum()
```

```python
#Excluding duplicated rows, how many works in df were written by Beethoven?
df.loc[~df.duplicated() & 
                 (df["Composer"] == "Ludwig van Beethoven")].shape[0]
```

```python
#The Romantic time period can roughly be defined by the years 1850-1920 inclusive. 
#Excluding duplicate rows, how many pieces were in this time frame?

df.loc[~df.duplicated() & 
                (df["Year"].between(1850,1920))].shape[0]
```

```python
#Including duplicates, how many pieces were written in a Major key?

df[df["Key"].str.contains("Major")].shape[0]

```

```python
# Including duplicates, how many pieces were written in a sharp key? Sharps can be denoted using the word "sharp" or "#".


df[df["Key"].str.contains("sharp|#", case=False)].shape[0]
```

```python
#How many different Time Signatures are present in df?


df["Time_Signature"].nunique()
```

```python
#Suppose you were to binge listen to every song in df including duplicates. 
#How many minutes would this take you assuming the data in the DataFrame is stored in minutes?

df["Length"].sum()
```

```python
#  Length of the piece with the largest number of movements

df.loc[df["Number_of_Movements"].idxmax(), "Length"]


```

```python
# Count of digits 1 and 2 in all titles

df["Title"].str.count("1|2").sum()
```

```python
#From the df_classical DataFrame, create a new DataFrame df_final where all data is the same,
#except the "Composer" column is changed so that it only includes the last name of each composer.

df_final = df.copy()
df_final["Composer"] = df_final["Composer"].str.split().str[-1]
```

```python
#Modify the given df DataFrame to include a column for Period corresponding to the table above for each work in the table. 
#Do the modification in place.


df.loc[df["Year"].between(1400, 1600), "Period"] = "Renaissance"
df.loc[df["Year"].between(1601, 1750), "Period"] = "Baroque"
df.loc[df["Year"].between(1751, 1820), "Period"] = "Classical"
df.loc[df["Year"].between(1821, 1900), "Period"] = "Romantic"
df.loc[df["Year"].between(1901, 2000), "Period"] = "20th Century"
df.loc[df["Year"] >= 2001, "Period"] = "Contemporary"
```
