```python
#MULTIPLE PLOTTING, 
#matplotlib will give us different colours for the different points or lines,
#depending on which you're plotting.

f = lambda x: x**2
g = lambda x: x 

x_val = np.array([1,2,3,4,5])

plt.scatter(x_val,f(x_val))
plt.scatter(x_val,g(x_val))
plt.scatter(x_val,g(x_val),color="g", marker="^")
plt.title('Simple Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
```

```python
import matplotlib.pyplot as plt
import numpy as np

##Setting up values on the x-axis
x = np.linspace(1, 5, 5, endpoint=True)

##Setting up values on the y-axis
f1 = lambda x: x**2
f2 = lambda x: x
f3 = lambda x: 30 - 5*x
f4 = lambda x: 6 - x
y1 = f1(x)
y2 = f2(x)
y3 = f3(x)
y4 = f4(x)

# Creates a figure with 2 rows and 2 columns (2x2 grid of subplots)
fig, axs = plt.subplots(2, 2, figsize=(6, 4))

#Add a bit of padding around the subplots.
plt.subplots_adjust(hspace=0.2, wspace=0.2)

# Can tighten the layout if desired using this:
#plt.tight_layout()

# First subplot (top left)
axs[0, 0].plot(x, y1, 'o')
axs[0, 0].set_title('Plot 1: y = x^2')
axs[0, 0].set_xlabel('x')
axs[0, 0].set_ylabel('y')

# Second subplot (top right)
axs[0, 1].plot(x, y2, 'o')
axs[0, 1].set_title('Plot 2: y = x')


axs[0, 1].set_xlabel('x')
axs[0, 1].set_ylabel('y')

# Third subplot (bottom left)
axs[1, 0].plot(x, y3, 'o')
axs[1, 0].set_title('Plot 3: y = 30 - x')
axs[1, 0].set_xlabel('x')
axs[1, 0].set_ylabel('y')

# Fourth subplot (bottom right)
axs[1, 1].plot(x, y4, 'o')
axs[1, 1].set_title('Plot 4: y = 6 - x')
axs[1, 1].set_xlabel('x')
axs[1, 1].set_ylabel('y')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np 

x = np.linspace(1, 5, 5, endpoint=True)
f1 = lambda x: x**2
f2 = lambda x: x
f3 = lambda x: 30 - 5*x
f4 = lambda x: 6 - x
y1 = f1(x)
y2 = f2(x)
y3 = f3(x)
y4 = f4(x)

# Creates a figure of width 10 and height 8
plt.figure(figsize=(10, 8))
#Add a bit of padding around the subplots.
plt.subplots_adjust(hspace=0.5, wspace=0.5)

# First subplot (top left)
plt.subplot(2, 2, 1)
plt.plot(x, y1, 'o')
plt.title('Plot 1: y = x^2')
plt.xlabel('x')
plt.ylabel('y')

# Second subplot (top right)
plt.subplot(2, 2, 2)
plt.plot(x, y2, 'o')
plt.title('Plot 2: y = 30 - x')
plt.xlabel('x')
plt.ylabel('y')

# Third subplot (bottom left)
plt.subplot(2, 2, 3)
plt.plot(x, y3, 'o')
plt.title('Plot 3: y = x')
plt.xlabel('x')
plt.ylabel('y')

# Fourth subplot (bottom right)
plt.subplot(2, 2, 4)
plt.plot(x, y4, 'o')
plt.title('Plot 4: y = 6 - x')
plt.xlabel('x')
plt.ylabel('y')

plt.show()
```
