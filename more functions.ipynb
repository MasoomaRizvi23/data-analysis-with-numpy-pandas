```python
df["Age"] = df["Age"].apply(lambda x: x+1)
print(df)
```

```python
#The code to create the new AgeGroup column. 

df["AgeGroup"] = df.apply(lambda row: 
                         ("young" if row["Age"] < 30
                          else "adult" if row["Age"] <=50
                          else "senior"), axis =1
                        )

#creating for income level
df["IncomeLevel"] = df.apply(
    lambda row: (
        "low earner" if row["Salary"] < 50000
        else "mid earner" if row["Salary"] <= 100000
        else "high earner"
    ),
    axis=1
)
```

```python
# 1. Percentage mapping 
df["salarypct"] = df["Salary"].map(lambda x : x/100000)
print(df)
```
