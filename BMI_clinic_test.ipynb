```python
import numpy as np

def bmi_categories(arr: np.ndarray[float]) -> np.ndarray[str]:
    """
  Returns an array of the same shape as arr
  but with each value replaced by a string corresponding
  to the classification of the number in the BMI scale.
  
  Requires: arr[i] > 0.0 for all indices i
  
  Examples:
     A = np.array([18.5, 24.0, 30.1, 29.5, 35.0, 22.4, 27.8])
     bmi_categories(A) => np.array(['Normal weight', 'Normal weight', 
                                    'Obese', 'Overweight', 'Obese', 
                                    'Normal weight', 'Overweight'])
     
     B = np.array([[13.5, 23.0], [30.1, 29.9]])
     bmi_categories(B) => np.array([['Underweight', 'Normal weight'],
                                    ['Obese', 'Overweight']])
  """

    result = np.empty(arr.shape, dtype=object)
 # Apply BMI category rules
    result[arr < 18.5] = "Underweight"
    result[(arr >= 18.5) & (arr < 25.0)] = "Normal weight"
    result[(arr >= 25.0) & (arr < 30.0)] = "Overweight"
    result[arr >= 30.0] = "Obese"

    return result

A = np.array([18.5, 24.0, 30.1, 29.5, 35.0, 22.4, 27.8])
print(bmi_categories(A))
```
