```python
import numpy as np

def cholesterol_classification(cholesterol_levels:
       np.ndarray[int|float]) -> np.ndarray[str]:
  """
  Returns a converted array where each value in cholesterol_levels
  is converted to a cholesterol classification
  
  Examples:
    cholesterol_classification(np.array([180, 220, 250])) => 
        np.array(['Normal', 'Borderline High', 'High'])
    cholesterol_classification(np.array([[240.0, 245.5], [250.9, 239.9]]) => 
        np.array([['High', 'High'], ['High', 'Borderline High']])
  """
  ##YOUR CODE GOES HERE
  conditions =[
    cholesterol_levels < 200,
        cholesterol_levels < 240
    ]
  choices = ['Normal', 'Borderline High']
  return np.select(conditions, choices, default='High')
# Test all boundary values at once
test_values = np.array([199.9, 200, 220, 239.9, 240, 250])
print(cholesterol_classification(test_values))

# Expected output:
# ['Normal' 'Borderline High' 'Borderline High' 'Borderline High' 'High' 'High']
pass
```
