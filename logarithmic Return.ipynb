```python
import numpy as np

def logarithmic_returns(stock_prices: np.ndarray[int|float]) \
      -> np.ndarray[float]:
    """
  Returns the logarithmic returns for the stock for consecutive days.
  
  Requires: 
     len(stock_prices.shape) == 1
     stock_prices[i] > 0 for all indices i
  
  Example:
     logarithmic_returns(np.array([100, 110, 120, 130]))
        => array([0.09531018, 0.08701138, 0.08004271])
"""
  #YOUR CODE GOES HERE
    ##adding edge case handling
    if stock_prices.ndim != 1:
        raise ValueError("stock_prices must be a 1‑D array")

    if np.any(stock_prices <= 0):
        raise ValueError("All stock prices must be positive")
    
    return np.log(stock_prices[1:] / stock_prices[:-1])
  
  #test
  
# Example input
prices = np.array([100, 110, 120, 130])

# Print the output
print(logarithmic_returns(prices))

pass
```
