```python


#======================================================================================

# Quick Memory Trick
#Bar plot = average height

#Count plot = how many

#Point plot = average as a dot

#======================================================================================
```

```python

#================================================================================================================================================
#| Plot           | Best For                          | Shows                     | Avoid When                                         |
#| -------------- | --------------------------------- | ------------------------- | --- -----------------------------------------------|
#| **Bar Plot**   | Comparing means across categories | Mean + error bar          | When you have many categories (bars get cluttered) |
#| **Count Plot** | Counting category frequencies     | Raw counts or percentages | When you need continuous variable summaries        |
#| **Point Plot** | Comparing means with trends       | Mean + CI as points       | When you want actual bar heights                   |

#================================================================================================================================================
```

```python
import matplotlib.pyplot as plt 
import pandas as pd 
import seaborn as sns 


penguins  = sns.load_dataset("penguins")
print(penguins)
```

```python
plt.figure(figsize=(6,10))
plt.subplot(2,1,1)
sns.barplot(penguins, x='species', y='body_mass_g')
plt.title("average body mass across species")

plt.subplot(2,1,2)
ax = sns.barplot(penguins, x='sex', y = 'bill_length_mm',estimator='mean',errorbar='sd',hue='sex', legend = False) 

#for color1 

ax.bar_label(ax.containers[0],fontsize = 10, label_type="center")
ax.bar_label(ax.containers[1],fontsize=10,label_type='center')

plt.title("aveage bill length across gender")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns


penguins  = sns.load_dataset("penguins")
plt.figure(figsize=(6,15))
plt.subplot(3,1,1)
sns.pointplot(penguins, x="species", y="body_mass_g")
plt.title("Average Body Mass Across Species")

plt.subplot(3,1,2)
#Remove line between points with linestyles
sns.pointplot(penguins, x="species", y="body_mass_g", linestyles='')
plt.title("Average Body Mass Across Species")

plt.subplot(3,1,3)
ax = sns.pointplot(penguins, x="species", y="bill_length_mm",
                 estimator = "mean", errorbar="sd",
                 linestyles = '',
                 hue="sex", legend=False, dodge=True)

plt.title("Average Bill Length Across Gender")
plt.show()
```
