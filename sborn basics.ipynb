```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

#----------we can execute these set of lines to see good set of data.......................
# pd.set_option('display.max_rows', 200)
# pd.set_option('display.max_columns', 200)
# pd.set_option('display.width', None)

iris = sns.load_dataset("iris")
iris.head()
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
sns.scatterplot(iris)

plt.subplot(2,2,2)
sns.barplot(iris)

plt.subplot(2,2,3)
sns.lineplot(iris)

plt.subplot(2,2,4)
sns.boxplot(iris)
```

```python
sns.pairplot(iris)
```

```python
sns.pairplot(iris, hue="species", markers=["o", "v", "D"])
```

```python
# small example of using kdeplot to estimate the distribution of the sepal length of the irises. An interesting activity is to do this for each of the three types of species.-------

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,10))
plt.axis('off')

plt.title('KDEs from Iris Dataset')

plt.subplot(3,1,1)
sns.kdeplot(iris['sepal_length'], fill=True)

#Remember when a single column works often multiple can as well in Seaborn!
plt.subplot(3,1,2)
sns.kdeplot(iris[iris.columns[:-1]], fill=True)

#Separating the KDEs by species:
plt.subplot(3,1,3)
sns.kdeplot(iris, x="sepal_length", hue="species", fill=True)

plt.subplots_adjust(hspace=0.5, wspace=0.5)


plt.show()
```
