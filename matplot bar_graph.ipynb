```python
import matplotlib.pyplot as plt
import pandas as pd  
dfg=pd.read_csv(r"C:\Users\masri\Downloads\7_3_dfg.csv")

value_counts = dfg.Grade.value_counts()

plt.figure(figsize=(100, 30))
value_counts.index = value_counts.index.astype(str)
plt.bar(value_counts.index, value_counts.values)
#Make the x-tick values larger and easier to read
plt.xticks(fontsize=50, rotation=45)
plt.show()

##--------------------no numeric values are in between, and sata index plot directly.  
```

```python
###use of sort_index

import matplotlib.pyplot as plt

value_counts = dfg.Grade.value_counts()

plt.figure(figsize=(100, 30))
value_counts.index = value_counts.index.astype(str)
value_counts.sort_index(inplace = True)
plt.bar(value_counts.index, value_counts.values)
plt.xticks(fontsize=50, rotation=45)
plt.show()
##------------100 comes far mroe left,
```

```python
#100 is on far right and non numeic are also at the end

import matplotlib.pyplot as plt

value_counts = dfg.Grade.value_counts()
value_counts.index = value_counts.index.astype(str)
value_counts.sort_index(inplace = True,
    key = lambda x: x.map(lambda x: str(x).zfill(3) 
                          if x.isnumeric() else x))
plt.figure(figsize=(100, 30))
plt.bar(value_counts.index.astype(str), value_counts.values)
#For horizontal bars, swap the above line with the line below.
#plt.barh(value_counts.index.astype(str), value_counts.values)
plt.xticks(fontsize=50, rotation=45)
plt.show()
```

```python
#here is exactly how to use "000" + x in your code so that DNW, INC, 
#and any other non‑numeric grades always appear on the left, 
#n alphabetical order, before the numeric grades.

import matplotlib.pyplot as plt

value_counts = dfg.Grade.value_counts()
value_counts.index = value_counts.index.astype(str)
value_counts.sort_index(inplace = True,
    key = lambda idx: idx.map(lambda x: str(x).zfill(3) 
                          if x.isnumeric() else "000"+x))
plt.figure(figsize=(100, 30))
plt.bar(value_counts.index.astype(str), value_counts.values)
#For horizontal bars, swap the above line with the line below.
#plt.barh(value_counts.index.astype(str), value_counts.values)
plt.xticks(fontsize=50, rotation=45)
plt.show()
```
