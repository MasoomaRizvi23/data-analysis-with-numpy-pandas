```python
import numpy as np

p = np.polynomial.Polynomial([1])
q = np.polynomial.Polynomial([1, 2])
r = np.polynomial.Polynomial([3, -2, 1])
print("p(x): ", p)
print("q(x): ", q)
print("r(x): ", r)
print("Notice the order of the coefficients is lowest first", r.coef)
#Note p.degree() is p.order with poly1d
print("Degree of p: ", p.degree()) 
print("q(-2): ", q(-2)) #Evaluate q at 2
print("p+q: ", p + q)
print("p-q: ", p - q)
print("q*r: ", q * r)
# Note that polydiv needs coefficients in the order of poly1d.
quo, rem = np.polydiv(r.coef[::-1],q.coef[::-1])
# Reverse them back to get np.polynomial.Polynomial objects.
quo = np.polynomial.Polynomial(quo[::-1])
rem = np.polynomial.Polynomial(rem[::-1])
print("Quotient: ", quo)
print("Remainder: ", rem)
print("Original: ", quo * q + rem, quo * q + rem == r)
# Find Roots of polynomials (values of x where q(x) = 0)
print("Roots of q: ", q.roots())
#Make a polynomial from the roots!
print("Polynomial from roots -1 and -2")
roots = [-1, -2]
coefficients = np.polynomial.polynomial.polyfromroots(roots)
print("Coefficients: ", coefficients)
print("Polynomial: ", np.polynomial.Polynomial(coefficients))

# Calculus: Derivative and Integrals. 
# Don't worry if you haven't seen Calculus yet!
# This is just for demonstration purposes if you have seen Calculus.
print("Derivative: ", q.deriv())
print("Integral: ", q.integ())
```

```python
#plot them using the plot method.

import numpy as np
import matplotlib.pyplot as plt

r = np.polynomial.Polynomial([3, -2, 1])
x = np.arange(-10, 10, 0.01)

plt.plot(x, r(x))
plt.show()
```

```python
# where we plot three polynomials and plot the sum of two of them on the same graph:

import numpy as np
import matplotlib.pyplot as plt

# Polynomial 1: p(x) = 2x^2 + 3x + 1
p1 = np.polynomial.Polynomial([1, 3, 2]) 

# Polynomial 2: q(x) = -x^2 + 4x - 2
p2 = np.polynomial.Polynomial([-2, 4, -1])

# Polynomial 3: r(x) = x^3 - 5x^2 + 4
p3 = np.polynomial.Polynomial([4, 0, -5, 1])

# Polynomial 4: p1 + p2
p4 = p1 + p2

x = np.linspace(-5, 5, 400)

y1 = p1(x)
y2 = p2(x)
y3 = p3(x)
y4 = p4(x)

plt.plot(x, y1, label='p(x) = 2x^2 + 3x + 1', color='blue')
plt.plot(x, y2, label='q(x) = -x^2 + 4x - 2', color='green')
plt.plot(x, y3, label='r(x) = x^3 - 5x^2 + 4', color='red')
plt.plot(x, y4, label='p(x) + q(x)', color='orange')

plt.title('Polynomials p(x), q(x), and r(x)')
plt.xlabel('x')
plt.ylabel('y')

plt.legend()

plt.show()
```
