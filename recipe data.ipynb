```python
import pandas as pd
df_recipes = pd.read_csv(r"C:\Users\masri\Downloads\7_0_df_recipes.csv")
df_ingredients = pd.read_csv(r"C:\Users\masri\Downloads\7_0_df_ingredients.csv")

df_recipes
```

```python
#How many int64 columns in df_recipes?


int64_col  = df_recipes.select_dtypes(include="int64").columns.tolist()
print(int64_col)
```

```python
#Q5 — Difficulty counts:

counts = df_recipes['difficulty'].value_counts()
print("Easy:", counts['Easy'])
print("Medium:", counts['Medium'])
print("Hard:", counts['Hard'])
```

```python
#— How many recipes use eggs?

eggs = df_ingredients[df_ingredients['ingredient'].str.lower() == 'eggs']['recipe_id'].nunique()
print(eggs)
```

```python
#What is the name of the recipe that takes the longest total time to make? Do not use string quotes in your answer!
df_recipes['total_time'] = df_recipes['prep_time_min'] + df_recipes['cook_time_min']
longest = df_recipes.loc[df_recipes['total_time'].idxmax(), 'title']
print(longest)
```

```python
# Author of the oldest recipe?
oldest_idx = pd.to_datetime(df_recipes['date_added']).idxmin()
print(df_recipes.loc[oldest_idx, 'author'])
```

```python
##max ingredient ijn singel recipe 
max_ing = df_ingredients.groupby('recipe_id').size().max()
print(max_ing)
```

```python
#Q9 — Max ingredients in a single recipe
max_ing = df_ingredients.groupby('recipe_id').size().max()
print(max_ing)
```

```python
#Q10 — Longest ingredient name
longest_ing = df_ingredients['ingredient'].str.len().idxmax()
print(df_ingredients.loc[longest_ing, 'ingredient'])
```

```python
#Q11 — Chef John's most common ingredient
chef_john_ids = df_recipes[df_recipes['author'] == 'Chef John']['recipe_id']
chef_john_ings = df_ingredients[df_ingredients['recipe_id'].isin(chef_john_ids)]
print(chef_john_ings['ingredient'].value_counts().idxmax())
```
