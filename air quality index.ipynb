```python
import numpy as np

def aqi(measurements: np.ndarray[float]) -> np.ndarray[str]:
  """
  Classifies the air quality index given the 6 readings per line
  
  Requires: 
     np.ndarray.shape == (n, 6) for some n.
     0.0 <= measurements[i] <= 500.0 for all indices i
  
  Example:
     aqi(np.array([[30.0, 20.0, 10.0, 5.4, 2.5, 10.2]])) =>
         np.array(["Good"])
  """
  ##Use these values in your answer given here to avoid spelling errors.
  possibilities = ["Good", "Moderate", "Unhealthy for Sensitive Groups",
                   "Unhealthy", "Very Unhealthy", "Hazardous"]
  # Step 1: compute AQI per row (max of the 6 pollutants)
  max_vals = np.max(measurements, axis=1)

    # Step 2: create output array
  result = np.empty(max_vals.shape, dtype=object)

    # Step 3: classify using masks
  result[max_vals <= 50] = possibilities[0]  # Good
  result[(max_vals > 50) & (max_vals <= 100)] = possibilities[1]  # Moderate
  result[(max_vals > 100) & (max_vals <= 150)] = possibilities[2]  # Sensitive
  result[(max_vals > 150) & (max_vals <= 200)] = possibilities[3]  # Unhealthy
  result[(max_vals > 200) & (max_vals <= 300)] = possibilities[4]  # Very Unhealthy
  result[(max_vals > 300) & (max_vals <= 500)] = possibilities[5]  # Hazardous

  return result
  pass
```
