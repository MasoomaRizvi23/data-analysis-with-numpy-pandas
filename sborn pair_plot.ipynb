```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

#================================================================================== 

#PAIRPLOT
#================================================================================== 


df_mpg  = sns.load_dataset("mpg")

sns.pairplot(df_mpg, 
             hue="origin",                
             diag_kind="kde",             
             markers=["o", "s", "D"],     
             palette="Set1",              
             height=2.5,   #Height of each facet (cell)                
             aspect=1.2,   #Controls width of each facet (cell)                
             plot_kws={'alpha': 0.7},      
             diag_kws={'fill': True})
```

```python
#================================================================================== 

#JOITN PLOT
#The marginal_kws dictionary controls the features of the side histogram plots,\
#whereas the joint_kws controls the features of the main grid.
#================================================================================== 

sns.jointplot(data=df_mpg, x="horsepower", y="mpg", 
              kind="reg",         
              color="red",          
              height=8,           
              ratio=5,             
              xlim=(50, 200),     
              ylim=(5, 50),
              joint_kws={'line_kws': {'color': 'green'}, 'ci':False},
              marginal_kws={'fill': False,'lw':5,'color':'blue'})
#can add 'kde':False to marginal_kws to make this false.
```

```python


#================================================================================== 

#FACET GRID
#================================================================================== 

g = sns.FacetGrid(df_mpg, row="origin", col="cylinders", 
                  hue="model_year")
g.map(sns.scatterplot, "acceleration", "mpg")
g.map(sns.histplot, "acceleration", kde=True)
g.add_legend()
g.set_axis_labels("Displacement (cu. in.)", "Miles per Gallon (mpg)")
g.set_titles("{row_name} - {col_name} Cylinders")
plt.show()
```
