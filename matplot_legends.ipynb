```python
import matplotlib.pyplot as plt
import numpy as np 

x = np.linspace(1,5,5, endpoint = True)

f1 = lambda x: x**2
f2 = lambda x: x
f3 = lambda x: 30 - 5*x
f4 = lambda x: 6 - x
y1 = f1(x)
y2 = f2(x)
y3 = f3(x)
y4 = f4(x)

fig, axs = plt.subplots(2,2,sharex = True, sharey = True, figsize=(6,5))
fig.suptitle("Assorted Plots")
fig.supxlabel("X")
fig.supylabel("Y")

plt.subplot(2,2,1)
plt.plot(x,y1,"go",label = "y=x^2")

plt.subplot(2,2,2)
plt.plot(x,y2,"b^",label ="y=x")

plt.subplot(2,2,3) 
plt.plot(x,y3,"yx",label = "y = 30-5x")

plt.subplot(2,2,4)
plt.plot(x,y4,"rv", label = "y = 6-x")


#-------------------------------------------------Add the legend to the right of the plot----------------------------------------------#

lines_labels = [ax.get_legend_handles_labels() for ax in axs.flat]
lines, labels = [sum(lol,[]) for lol in zip(*lines_labels)]
fig.legend(lines, labels, loc = "lower right")

#Add a bit of room to fit the legend.
fig.subplots_adjust(right = 0.7)
plt.show()
```
