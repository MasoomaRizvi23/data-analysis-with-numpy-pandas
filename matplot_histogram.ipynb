```python
import matplotlib.pyplot as plt
import numpy as np

import pandas as pd

data = {
    "Name": [
        "Andrew Gonzales", "Mark Jenkins", "Sara Sullivan", 
        "Gene Allen", "Adam Curry",
        # ... continue until row 1056
        "Jennifer Reed", "Joseph Wilkins", "Caitlin Johnson",
        "Sarah Dunlap", "Emily Williams"
    ],
    "Grade": [
        32.488, 87.574, 48.598, 73.315, 84.001,
        # ... continue numeric grades ...
        "INC", "INC", "INC", "INC", "INC"
    ]
}

dfgf = pd.DataFrame(data)
```

```python
import matplotlib.pyplot as plt
import numpy as np

grades = pd.to_numeric(dfgf.Grade, errors='coerce').dropna()

plt.hist(grades, bins=7, edgecolor='black')

plt.xlabel('Grade')
plt.ylabel('Frequency')
plt.title('Histogram of Grades')

plt.show()
```

```python
#need to set the bins parameter to be a sequence of the bin values.

import matplotlib.pyplot as plt
import numpy as np

bin_edges = [0, 50, 60, 70, 80, 100]
grades = pd.to_numeric(dfgf.Grade, errors='coerce').dropna()

plt.hist(grades, bins=bin_edges, edgecolor='black')

plt.xlabel('Grade')
plt.ylabel('Frequency')
plt.title('Histogram of Grades')

plt.show()
```

```python
#we can choose to make the bars the same size by setting the width parameter but...abs

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

grades = pd.to_numeric(dfgf.Grade, errors='coerce').dropna()

bin_edges = [0, 50, 60, 70, 80, 100]

#Lets use letter grades for the bin values
plt.xticks(np.array(bin_edges[:-1])+4, ['F', 'D', 'C', 'B', 'A'])  
plt.hist(grades, bins=bin_edges, edgecolor='black', width=8)

plt.xlabel('Grade Range')
plt.ylabel('Frequency')
plt.title('Histogram of Grades')

plt.show()
```
